```python
from fpdf import FPDF
from datetime import datetime
from IPython.display import clear_output
from google.colab import files

contatos = []
arquivo_txt = "contatos_agenda.txt"

def salvar_contatos():
    with open(arquivo_txt, "w", encoding="utf-8") as f:
        for c in contatos:
            f.write(f"{c['nome']};{c['telefone']}\n")

def backup_contatos():
    agora = datetime.now().strftime("%Y-%m-%d_%H-%M")
    nome_backup = f"contatos_backup_{agora}.txt"
    with open(nome_backup, "w", encoding="utf-8") as f:
        for c in contatos:
            f.write(f"{c['nome']};{c['telefone']}\n")
    print(f"[✓] Backup criado: {nome_backup}")

def exportar_para_pdf():
    if not contatos:
        print("Nenhum contato para exportar.")
        return
    pdf = FPDF()
    pdf.add_page()
    pdf.set_font("Arial", size=12)
    pdf.cell(200, 10, txt="Agenda de Contatos", ln=True, align="C")
    pdf.ln(10)
    for c in contatos:
        linha = f"Nome: {c['nome']}    Telefone: {c['telefone']}"
        pdf.cell(200, 10, txt=linha, ln=True)
    nome_pdf = "contatos_agenda.pdf"
    pdf.output(nome_pdf)
    print("[✓] PDF gerado com sucesso!")
    files.download(nome_pdf)

def listar_contatos():
    if not contatos:
        print("Nenhum contato cadastrado.")
        return
    for i, c in enumerate(contatos, 1):
        print(f"{i}. {c['nome']} - {c['telefone']}")

def buscar_contato():
    termo = input("Digite o nome para buscar: ").lower()
    achou = False
    for i, c in enumerate(contatos, 1):
        if termo in c['nome'].lower():
            print(f"{i}. {c['nome']} - {c['telefone']}")
            achou = True
    if not achou:
        print("Nenhum contato encontrado com esse nome.")

def editar_contato():
    listar_contatos()
    try:
        idx = int(input("Número do contato para editar: ")) - 1
        if 0 <= idx < len(contatos):
            novo_nome = input(f"Novo nome (atual: {contatos[idx]['nome']}): ")
            novo_tel = input(f"Novo telefone (atual: {contatos[idx]['telefone']}): ")
            contatos[idx]['nome'] = novo_nome or contatos[idx]['nome']
            contatos[idx]['telefone'] = novo_tel or contatos[idx]['telefone']
            salvar_contatos()
            backup_contatos()
            print("[✓] Contato editado com sucesso!")
        else:
            print("Índice inválido.")
    except ValueError:
        print("Entrada inválida.")

def remover_contato():
    listar_contatos()
    try:
        idx = int(input("Número do contato para remover: ")) - 1
        if 0 <= idx < len(contatos):
            removido = contatos.pop(idx)
            salvar_contatos()
            backup_contatos()
            print(f"[✓] Contato '{removido['nome']}' removido com sucesso!")
        else:
            print("Índice inválido.")
    except ValueError:
        print("Entrada inválida.")

def menu():
    while True:
        print("\n==== AGENDA DE CONTATOS ====")
        print("1. Adicionar contato")
        print("2. Listar contatos")
        print("3. Buscar por nome")
        print("4. Editar contato")
        print("5. Remover contato")
        print("6. Exportar para PDF")
        print("7. Baixar arquivo .txt")
        print("8. Sair")
        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            nome = input("Nome: ")
            telefone = input("Telefone: ")
            contatos.append({'nome': nome, 'telefone': telefone})
            salvar_contatos()
            backup_contatos()
            clear_output()
            print("[✓] Contato salvo com sucesso!")

        elif opcao == "2":
            listar_contatos()

        elif opcao == "3":
            buscar_contato()

        elif opcao == "4":
            editar_contato()

        elif opcao == "5":
            remover_contato()

        elif opcao == "6":
            exportar_para_pdf()

        elif opcao == "7":
            files.download(arquivo_txt)

        elif opcao == "8":
            print("Encerrando... Obrigado por usar a agenda!")
            break

        else:
            print("Opção inválida. Tente novamente.")

# Iniciar a Agenda
menu()
```
