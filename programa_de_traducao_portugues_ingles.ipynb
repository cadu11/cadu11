```python
import sys
from PyQt5.QtWidgets import (
    QApplication, QWidget, QLabel, QLineEdit,
    QPushButton, QVBoxLayout, QComboBox, QMessageBox
)
from PyQt5.QtGui import QFont
from PyQt5.QtCore import Qt
from googletrans import Translator
import pyttsx3

class TradutorApp(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Tradutor PT ⇄ EN com Voz")
        self.setGeometry(300, 300, 400, 250)
        self.setStyleSheet("background-color: #f0f8ff;")

        self.layout = QVBoxLayout()

        self.titulo = QLabel("🔤 Tradutor de Português ⇄ Inglês")
        self.titulo.setFont(QFont("Arial", 16, QFont.Bold))
        self.titulo.setAlignment(Qt.AlignCenter)
        self.layout.addWidget(self.titulo)

        self.combo = QComboBox()
        self.combo.addItems(["Português → Inglês", "Inglês → Português"])
        self.combo.setFont(QFont("Arial", 11))
        self.layout.addWidget(self.combo)

        self.entrada = QLineEdit()
        self.entrada.setPlaceholderText("Digite o texto para traduzir...")
        self.entrada.setFont(QFont("Arial", 12))
        self.layout.addWidget(self.entrada)

        self.botao = QPushButton("Traduzir e Falar")
        self.botao.setStyleSheet("background-color: #007acc; color: white; font-weight: bold;")
        self.botao.clicked.connect(self.traduzir)
        self.layout.addWidget(self.botao)

        self.resultado = QLabel("")
        self.resultado.setFont(QFont("Arial", 14))
        self.resultado.setAlignment(Qt.AlignCenter)
        self.layout.addWidget(self.resultado)

        self.setLayout(self.layout)

    def traduzir(self):
        texto = self.entrada.text()
        if not texto.strip():
            QMessageBox.warning(self, "Aviso", "Digite uma frase ou palavra.")
            return

        direcao = self.combo.currentText()
        src_lang = 'pt' if "Português" in direcao else 'en'
        dest_lang = 'en' if src_lang == 'pt' else 'pt'

        try:
            tradutor = Translator()
            traducao = tradutor.translate(texto, src=src_lang, dest=dest_lang)
            self.resultado.setText(f"Tradução: {traducao.text}")
            self.falar(traducao.text, dest_lang)
        except Exception as e:
            QMessageBox.critical(self, "Erro", "Erro ao traduzir. Verifique sua conexão com a internet.")

    def falar(self, texto, idioma):
        engine = pyttsx3.init()
        voices = engine.getProperty('voices')
        if idioma == 'pt':
            for voice in voices:
                if "brazil" in voice.name.lower() or "portuguese" in voice.name.lower():
                    engine.setProperty('voice', voice.id)
                    break
        elif idioma == 'en':
            for voice in voices:
                if "english" in voice.name.lower():
                    engine.setProperty('voice', voice.id)
                    break
        engine.say(texto)
        engine.runAndWait()

if __name__ == "__main__":
    app = QApplication(sys.argv)
    janela = TradutorApp()
    janela.show()
    sys.exit(app.exec_())
```
